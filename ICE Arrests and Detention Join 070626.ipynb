### In this notebook I will:
1. Pull the clean filtered data from arrests (that has removed duplicates and blank identifiers)
2. Use the unique identifiers there to cut through the raw parquet data of detentions
3. Join the clean filtered arrests data and the new filtered detentions data on the unique identifier data

```python
import pandas as pd
```

```python
##pulling in the data we got into this notebook
df = pd.read_csv('/Users/sajinashrestha/Documents/dataProjects/WAMC/ICEpresense-CapitalRegion/FilteredData/2025NYArrests_070626.csv', low_memory=False)
df
```

```python
df_UI = df['unique_identifier']
df_UI
```

```python
df_UI2 = df_UI.to_frame()
df_UI2
```

```python
df_RAWDetentions = pd.read_parquet('/Users/sajinashrestha/Documents/dataProjects/WAMC/ICEpresense-CapitalRegion/RawData/detention-stays-latest.parquet')
df_RAWDetentions
```

```python
##now wwe ask the notebook to use these unique identifiers to pull only those from the detentions list

filtered_detentions = df_RAWDetentions[df_RAWDetentions['unique_identifier'].isin(df_UI2['unique_identifier'])]
filtered_detentions
```

```python
##merge the two dataframes on unique identifier
ArrestsDetentions = pd.merge(df, filtered_detentions, on="unique_identifier")
ArrestsDetentions
```

```python
ArrestsDetentions['unique_identifier'].value_counts()
```

```python
ArrestsDetentions.to_csv('Filtereddata/2025ArrestsDetentions_070726.csv', index = False)
```
