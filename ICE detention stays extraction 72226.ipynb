```python
##pulling in the data we got into this notebook
df = pd.read_parquet(r"C:\Users\sshrestha\Desktop\Writing\DataProjects\ICE Deportations and Arrests\2024 Data\RAWDATA\detention-stays-latest(1).parquet")
df
```

```python
from datetime import datetime
```

```python
##turning stay in book time to datetime so we can filter an do equations to it
df["stay_book_in_date_time"] = pd.to_datetime(df["stay_book_in_date_time"])
```

```python
##whittling it to just 2025. this works because, factually, if someone was arrested in 2024, their detention stays
##following that arrest is not really going to be before the date. arrests LEAD to detentions

dfDetentions = df.loc[
  (df["stay_book_in_date_time"] >= "2024-01-01") &
  (df["stay_book_in_date_time"] <= "2024-12-31")] 
```

```python
dfDetentions
```

```python
##bringing in the unique identifiers from the arrests list so we can use that as a way to whittle this listfurther
dfArrestIDs = pd.read_csv(r"C:\Users\sshrestha\Desktop\Writing\DataProjects\ICE Deportations and Arrests\2024 Data\FilteredData\IDsNY2024Arrests_072226.csv")
dfArrestIDs
```

```python
##now wwe ask the notebook to use these unique identifiers to pull only those from the detentions list
filtered_detentions = dfDetentions[dfDetentions['unique_identifier'].isin(dfArrestIDs['unique_identifier'])]
filtered_detentions
```

```python
filtered_detentions.to_csv('Filtereddata/2024DetentionsofNYArrests_072226.csv', index = False)
```
