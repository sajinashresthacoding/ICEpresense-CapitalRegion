```python
##let's bring in the raw data to the front
df = pd.read_parquet(r'C:\Users\sshrestha\Desktop\Writing\DataProjects\ICE Deportations and Arrests\2024 Data\RAWDATA\arrests-latest.parquet')
df
```

```python
from datetime import datetime
```

```python
#let's turn the apprehension date, the dat that will tell us when someone was arrested
#into a date format so we can use filters and other formulas on it. 

df["apprehension_date"] = pd.to_datetime(df["apprehension_date"])
```

```python
##now let's use that apprehension date and state columns to pull only arrests made in 2024 within NYS

dfArrests = df.loc[
  (df["apprehension_date"] >= "2024-01-01") &
  (df["apprehension_date"] <= "2024-12-31") &
  (df["apprehension_state_filled_in"] == "NEW YORK")
]

dfArrests
```

```python
##let's save those arrests
dfArrests.to_csv('FilteredData/2024NYArrests_072226.csv', index = False)
```

```python
##we need to grab the unique identifier because that is how we will grab what we need from the detentions sheet
##it works as an "index" of sorts
dfIDsArrests = dfArrests[['unique_identifier']]
```

```python
dfIDsArrests.to_csv('Filtereddata/IDsNY2024Arrests_072226.csv', index = False)
```
