```python
##pulling in the data we got into this notebook
df = pd.read_csv(r"C:\Users\sshrestha\Desktop\Writing\DataProjects\ICE Deportations and Arrests\2024 Data\FilteredData\2024NYArrests_072226.csv", low_memory=False)
df
```

```python
df_UI = df['unique_identifier']
df_UI
```

```python
df_UI2 = df_UI.to_frame()
df_UI2
```

```python
df_RAWDetentions = pd.read_parquet(r"C:\Users\sshrestha\Desktop\Writing\DataProjects\ICE Deportations and Arrests\2024 Data\RAWDATA\detention-stays-latest(1).parquet")
df_RAWDetentions
```

```python
##now wwe ask the notebook to use these unique identifiers to pull only those from the detentions list

filtered_detentions = df_RAWDetentions[df_RAWDetentions['unique_identifier'].isin(df_UI2['unique_identifier'])]
filtered_detentions
```

```python
##merge the two dataframes on unique identifier
ArrestsDetentions = pd.merge(df, filtered_detentions, on="unique_identifier")
ArrestsDetentions
```
